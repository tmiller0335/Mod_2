```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
sns.get_dataset_names()
```

```python
df = sns.load_dataset('tips')
df.head()
```

```python
#normal Groupby
df.groupby('time')['total_bill'].mean()
```

```python
#.to_frame
df.groupby('time')['total_bill'].mean().to_frame('avg_bill')
```

```python
#.reset_index() - adds the index bar back
df.groupby('time')['total_bill'].mean().to_frame('avg_bill').reset_index()
```

```python
#.sort_values()
df.groupby('time')['total_bill'].mean().to_frame('avg_bill').reset_index().sort_values('avg_bill', ascending=False)
```

```python
#assign to new dataframe
avgbill = df.groupby('time')['total_bill'].mean().to_frame('avg_bill').reset_index().sort_values('avg_bill', ascending=False)
```

```python
avgbill.head()
```

```python
#common calculations .sum() / .count() / .min() / .Max() / .mean() 
df.groupby('time')['total_bill'].count()
#.count does NOT count nulls 
```

```python
df.groupby('time')['total_bill'].size()
#.size counts nulls 
```

```python
df.groupby('time')['total_bill'].nunique()
#nunique will only pull unique items, does not include duplicates
```

```python
df.groupby('time')['total_bill'].min()
```

```python
df.groupby('time')['total_bill'].max()
```

```python
df.groupby('time')['total_bill'].sum()
```

```python
df.groupby('time')['total_bill'].median()
```

```python
#Question 1# Find the AVG Tip fpr each night of the week.
df.groupby('day')['tip'].mean()
```

```python
#Question 2# Find the Max tip for each sex

df.groupby('time')['total_bill'].max()
```

```python
#Question 3# ->Quantile --> Top 10 % of Tips each Night
df.groupby('day')['tip'].quantile(.9)
```

```python
#multiple caculations .agg() + formats

df.groupby('time')['total_bill'].agg(['min', 'max'])
```

```python
df.groupby('time')['total_bill'].agg([min, max])
```

```python
df.groupby('time')['total_bill'].agg(['min', np.median, 'max'])
```

```python
df.groupby('time')['total_bill'].agg(minimum ='min', median = np.median, maximum = 'max')
```

```python
df.groupby('time')['total_bill'].agg(['min', np.median, 'max']).rename(columns = {'min': 'minimum', 'max': 'max'})
```

```python
df.groupby('time')['total_bill'].describe()
```

```python
df.groupby(['time', 'day'])[['tip','total_bill']].mean()
```
