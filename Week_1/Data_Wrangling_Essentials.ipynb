As you clean and prepare your dataset, you'll often need to rename confusing column names or reorder them for better readability or consistency. This is especially important when you're getting ready to share your dataset with others or use it in reports and visualizations.

You can rename a single column, or multiple columns at once (using a dictionary):

```python
df.rename(columns={'old_name': 'new_name'}, inplace=True)
df.rename(columns={
    'Dept': 'Department',
    'Sal': 'Salary'
}, inplace=True)
```

To change the column order, you can directly type in a new list of names for the columns, or dynamically using indexing:

```python
df = df[['Salary', 'Name', 'Department']]
cols = df.columns.tolist()
new_order = [cols[-1]] + cols[:-1]  # Moves the last column to the front
df = df[new_order]
```
