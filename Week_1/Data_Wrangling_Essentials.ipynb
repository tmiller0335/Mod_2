When you're working with a dataset, you often want to break it into groups and then calculate some statistics within each group. That’s exactly what groupby() lets you do.

Imagine you have a big table of sales data. You want to know:

What’s the total revenue per region?
What’s the average sale price per salesperson?
How many products were sold per category?
All of these are grouped calculations — and Pandas makes them easy.

```python
df.groupby('column_to_group_by')['column_to_aggregate'].agg_function()
```

```python
# Average salary per Department
df.groupby('Department')['Salary'].mean()
#  Average salary by Department and Role
df.groupby(['Department', 'Role'])['Salary'].mean()
```

So far, we’ve seen how .groupby() can be used to group rows by a category and calculate a single summary statistic — like .mean().

That’s a great start! But real-world analysis often requires more than one summary at a time — maybe you want to see both the average and the total, or the count and the maximum, or even different stats for different columns.

That’s where .agg() comes in.

.agg() (short for aggregate) lets you:

Apply multiple functions at once
Specify different functions for different columns
Add custom labels to make the results easier to read

```python
data = {
    'Department': ['HR', 'HR', 'IT', 'IT', 'Sales', 'Sales'],
    'Salary': [50000, 52000, 60000, 63000, 45000, 47000],
    'Tenure': [2, 3, 5, 4, 1, 2]
}
df = pd.DataFrame(df)

# Group by Department, get average and max salary
df.groupby('Department')['Salary'].agg(['mean', 'max'])
```

Let’s say you want to find the mean Salary and the max Tenure for each department. You can use dictionaries within .agg():

```python
df.groupby('Department').agg({
    'Salary': 'mean',
    'Tenure': 'max'
})
```

You can also rename the output columns to make them more readable using .agg() with tuples:

```python
df.groupby('Department').agg(
    Avg_Salary=('Salary', 'mean'),
    Max_Tenure=('Tenure', 'max')
)
```

```python
data = {
    'Store': ['A', 'A', 'A', 'B', 'B', 'C', 'C', 'C', 'C'],
    'Product': ['Apples', 'Oranges', 'Bananas', 'Apples', 'Bananas', 'Oranges', 'Apples', 'Bananas', 'Bananas'],
    'Sales': [100, 80, 90, 120, 110, 60, 95, 85, 75],
    'Profit': [30, 25, 20, 40, 35, 15, 32, 22, 18]
}

df = pd.DataFrame(data)

# View the first few rows
print(df.head())
```

```python
# 1. Group the data by Store and use .agg() to find the average Sales and average Profit for each store
df.groupby('Store').agg({
    'Sales' : 'mean',
    'Profit' : 'mean'
})
```

```python
# 2. Group by Product and get the maximum Sales and minimum Profit
df.groupby('Product')({
    'Sales' : 'max'
    'Profit' : 'min'
})
```
