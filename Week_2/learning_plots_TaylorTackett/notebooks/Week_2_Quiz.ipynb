Creating a Histogram

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "duration_ms": [210000, 180000, 240000, 300000, 190000, 220000, 260000, 250000]
})
df
```

```python
# Write code below to create and display a histogram showing the distribution of song durations
# Use 5 bins, 
# add a title "Distribution of Song Durations"
# label the axes "Duration" (x) and "Count" (y)

plt.hist(df['duration_ms'], bins=5)

plt.xlabel("Duration")
plt.ylabel("Count")
plt.title("Distribution of Song Durations")

plt.show()
```

Creating a Scatterplot

```python
df = pd.DataFrame({
    "energy": [0.6, 0.8, 0.5, 0.9, 0.7, 0.65],
    "valence": [0.5, 0.7, 0.4, 0.8, 0.6, 0.55]
})
df

# Write code below to create and display a scatter plot of energy vs valence
# Label the axes "Energy" (x) and "Valence" (y)
# Title the chart "Energy vs Valence"
```

```python
plt.scatter(df["energy"], df["valence"])

plt.xlabel("Energy")
plt.ylabel("Valence")
plt.title("Energy vs Valence")

plt.show()
```

Creating a Bar Plot:

```python
data = {
    "genre": ["Pop", "Rock", "Hip-Hop", "Electronic"],
    "streams": [120, 90, 150, 70]
}

df = pd.DataFrame(data)

# Write code below to create and display a bar plot showing streams by genre
# Label the axes "Genre"(x) and "Streams"(y)
# add a title "Streams by Genre"
```

```python
plt.bar(df["genre"], df["streams"])

plt.xlabel("Genre")
plt.ylabel("Streams")
plt.title("Streams by Genre")

plt.show()
```
